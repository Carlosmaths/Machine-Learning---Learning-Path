#Scatter plots sencillos

```python
N = 2000
random_x = np.random.randn(N)
random_y = np.random.randn(N)
```

```python
trace = go.Scatter(x = random_x, y = random_y, mode = 'markers')
```

```python
fig = go.Figure(data=[trace])
fig.update_layout(
    title="Gráfico de dispersión con 2000 puntos",
    xaxis_title="Eje X",
    yaxis_title="Eje Y",
    xaxis=dict(
        zeroline = True,  # Hacer visible el eje X en 0
        zerolinecolor = 'black',  # Color de la línea del eje X
        zerolinewidth = 2  # Grosor de la línea del eje X
    ),
    yaxis = dict(
        zeroline = True,  # Hacer visible el eje Y en 0
        zerolinecolor = 'black',  # Color de la línea del eje Y
        zerolinewidth = 2  # Grosor de la línea del eje Y
    ),
    template = 'plotly_white'  # Fondo blanco para que los ejes resalten
)

fig.show()
```

#Gráficos combinados

```python
N = 200
rand_x = np.linspace(0,1,N)
rand_y0 = np.random.randn(N) + 3
rand_y1 = np.random.randn(N)
rand_y2 = np.random.randn(N) - 3
```

```python
trace0 = go.Scatter(x = rand_x, y = rand_y0, mode = 'markers', name = 'Puntos')
trace1 = go.Scatter(x = rand_x, y = rand_y1, mode = 'lines', name = 'Lineas')
trace2 = go.Scatter(x = rand_x, y = rand_y2, mode = 'markers + lines', name = 'Puntos y Lineas')
data = [trace0, trace1, trace2]
```

```python
fig = go.Figure(data)
fig.update_layout(
    title="scatter-line-plot",
    xaxis_title="Eje X",
    yaxis_title="Eje Y",
    xaxis=dict(
        zeroline = True,  # Hacer visible el eje X en 0
        zerolinecolor = 'black',  # Color de la línea del eje X
        zerolinewidth = 2  # Grosor de la línea del eje X
    ),
    yaxis = dict(
        zeroline = True,  # Hacer visible el eje Y en 0
        zerolinecolor = 'black',  # Color de la línea del eje Y
        zerolinewidth = 2  # Grosor de la línea del eje Y
    ),
    template = 'plotly_white'  # Fondo blanco para que los ejes resalten
)

fig.show()
```

#Estilizado de gráficos

```python
trace = go.Scatter(x = random_x, y = random_y, mode = 'markers', name = 'Puntos con buen estilo',
                   marker = dict(size = 12, color = 'rgba(140,20,20,0.8)', line = dict(width = 2, color = 'rgb(10,10,10)')))
```

```python
layout = dict(title = 'Scatter Plot Estilizado', xaxis = dict(zeroline = False), yaxis = dict(zeroline = False))
```

```python
fig =  go.Figure(data = [trace], layout = layout)
fig.show()
```

```python
trace = go.Scatter(x = random_x, y = random_y, mode = 'markers', name = 'Puntos con buen estilo',
                   marker = dict(size = 8, color = 'rgba(10,80,220,0.25)', line = dict(width = 1, color = 'rgb(10,10,80)')))

layout = dict(title = 'Scatter Plot Estilizado', xaxis = dict(zeroline = False), yaxis = dict(zeroline = False))

fig =  go.Figure(data = [trace], layout = layout)
fig.show()
```

```python
trace = go.Histogram(x = random_x, name = 'Puntos con buen estilo')

layout = dict(title = 'Scatter Plot Estilizado', xaxis = dict(zeroline = False), yaxis = dict(zeroline = False))

fig =  go.Figure(data = [trace], layout = layout)
fig.show()
```

```python
trace = go.Box(x = random_x, name = 'Puntos con buen estilo')

layout = dict(title = 'Scatter Plot Estilizado', xaxis = dict(zeroline = False), yaxis = dict(zeroline = False))

fig =  go.Figure(data = [trace], layout = layout)
fig.show()
```

```python
import pandas as pd
```

```python
data = pd.read_csv('/content/drive/MyDrive/Machine Learning - Python/Datasets - Machine Learning/usa-population/usa_states_population.csv')
```

```python
data
```

```python
!pip install plotly
import plotly.graph_objects as go
import colorsys
```

```python
import numpy as np
N = len(data)
```

```python
# Generate list of N hls colors and convert them to rgb
c = [colorsys.hls_to_rgb(h / 360.0, 0.5, 0.5) for h in np.linspace(0, 360, N)]
# Convert rgb to 'rgb(r, g, b)' strings
c = ['rgb(' + ','.join(str(int(x * 255)) for x in color) + ')' for color in c]
```

```python
l = []
y = []

for i in range(N):
  y.append((2000 + i))
  trace0 = go.Scatter(
      x = data['Rank'],
      y = data['Population'] + i*1000000,
      mode = 'markers',
      marker = dict(size = 14, line = dict(width = 1), color = c[i]), opacity = 0.3,
      name = data['State'][i]
  )
  l.append(trace0)
```

```python
layout = go.Layout(title = 'Población de los estados de USA',
                   hovermode = 'closest',
                   xaxis = dict(title = 'Rango', ticklen = 5, zeroline = False, gridwidth = 2),
                   yaxis = dict(title = 'Población', ticklen = 5, gridwidth = 2),
                   showlegend = False)
```

```python
fig = go.Figure(data = l, layout = layout)
fig.show()
```
