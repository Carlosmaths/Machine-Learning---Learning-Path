```python
df = pd.read_csv('/content/drive/MyDrive/Machine Learning - Python/Datasets - Machine Learning/iris/iris.csv')
```

```python
X = df.iloc[:, 0: 4].values
y = df.iloc[:, 4].values
X_std = StandardScaler().fit_transform(X)
```

```python
from sklearn.decomposition import PCA as sk_pca
```

```python
pca = sk_pca(n_components = 2)
Y = pca.fit_transform(X_std)
```

```python
from plotly.graph_objs import Line, Marker, Scene, Figure
import plotly.offline as py
import plotly.graph_objs as go
```

```python
results = []

for name in ('setosa', 'versicolor', 'virginica'):
    result = go.Scatter(x = Y[y == name, 0], y = Y[y == name, 1],
                        mode = 'markers', name = name,
                        marker = Marker(size = 8, line = Line(color = 'rgba(220, 220, 220, 0.15)', width = 0.5),
                        opacity = 0.8))
    results.append(result)

data = Data(results)
layout = Layout(showlegend = True,
                scene = Scene(xaxis = XAxis(title = 'Componente Principal 1', showline = False),
                              yaxis = YAxis(title = 'Componente Principal 2', showline = False)))

fig = Figure(data = data, layout = layout)
py.iplot(fig)
```
