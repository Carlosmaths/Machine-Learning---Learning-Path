##1. Realizamos el cálculo de los valores y vectores propios.

####a) Usando la matriz de covarianzas.

```python
import numpy as np
```

```python
mean_vect = np.mean(X_std, axis = 0)
mean_vect
```

```python
cov_matrix = (X_std - mean_vect).T.dot((X_std - mean_vect)) / (X_std.shape[0] - 1)
cov_matrix
```

```python
np.cov(X_std.T)
```

```python
eig_vals, eig_vectors = np.linalg.eig(cov_matrix)
```

```python
print('Valores propios:')
eig_vals
```

```python
print('Vectores propios:')
eig_vectors
```

###b) Usando la matriz de correlaciones.

```python
corr_matrix = np.corrcoef(X_std.T)
corr_matrix
```

```python
'''
Notemos que el resltado siguiente es igual al anterior.
No hay necesidad de usar X_std, sino sólo X.
'''

corr_matrix = np.corrcoef(X.T)
corr_matrix
```

##2. Las Componentes Principales.

```python
for ev in eig_vectors:
    print('La longitud del vector propio es:' , np.linalg.norm(ev))
```

```python
eigen_pairs = [(np.abs(eig_vals[i]), eig_vectors[:, i]) for i in range(len(eig_vals))]
eigen_pairs
```

```python
eigen_pairs.sort()
eigen_pairs
```

###Ordenamos los vectores propios con valores propios de mayor a menor.

```python
eigen_pairs.reverse()
eigen_pairs
```

```python
print('Valores propios en orden descendente:')
print('')
for ep in eigen_pairs:
    print(ep[0])
```

```python
total_sum = sum(eig_vals)
total_sum
```

```python
'''
La varianza explicativa nos dice cuanta información,
cuanta varianza explica cada una de las componentes.
'''
var_exp = [(i / total_sum) * 100 for i in sorted(eig_vals, reverse = True)]
var_exp
```

```python
cum_var_exp = np.cumsum(var_exp)
cum_var_exp
```

```python
!pip install bokeh
from bokeh.plotting import figure, show, output_notebook
from bokeh.models import ColumnDataSource
import numpy as np

output_notebook()

# Datos ficticios para el ejemplo
var_exp = [40, 30, 20, 10]  # Porcentaje de varianza explicada por cada componente
cum_var_exp = np.cumsum(var_exp)  # Porcentaje acumulado

# Crear una ColumnDataSource para el gráfico de barras
source_bar = ColumnDataSource(data=dict(x=['CP %s' % i for i in range(1, 5)], y=var_exp))

# Crear una ColumnDataSource para el gráfico de dispersión y líneas
source_scatter = ColumnDataSource(data=dict(x=['CP %s' % i for i in range(1, 5)], y=cum_var_exp))

# Crear la figura principal
plot1 = figure(
    title="Porcentaje de variabilidad explicada por cada componente principal",
    x_axis_label="Componentes principales",
    y_axis_label="Porcentaje de varianza explicada",
    x_range=['CP %s' % i for i in range(1, 5)]
)

# Agregar barras verticales
plot1.vbar(x='x', top='y', width=0.5, source=source_bar, legend_label="Varianza Explicada")

# Agregar los puntos del gráfico de dispersión
plot1.circle(
    x='x', y='y', size=8, source=source_scatter, legend_label="Varianza Explicada Acumulada", color="red"
)

# Agregar líneas que conectan los puntos
plot1.line(
    x='x', y='y', source=source_scatter, legend_label="Línea Acumulada", color="red", line_width=2
)

# Personalizar la leyenda
plot1.legend.location = "top_right"  # Ubicar la leyenda
plot1.legend.click_policy = "hide"  # Permitir ocultar datos al hacer clic en la leyenda

# Mostrar el gráfico
show(plot1)
```

###Podemos seleccionar las dos primeras columnas, ya que ellas explican el 96% aproximadamente.

```python
w = np.hstack((eigen_pairs[0][1].reshape(4, 1),
              eigen_pairs[1][1].reshape(4, 1)))
w
```

###Proyectando las variables del nuevo espacio vectorial

```python
Y = X_std.dot(w)
Y
```

```python
from plotly.graph_objs import Line, Marker, Scene, Figure
import plotly.offline as py

results = []

for name in ('setosa', 'versicolor', 'virginica'):
    result = go.Scatter(x = Y[y == name, 0], y = Y[y == name, 1],
                        mode = 'markers', name = name,
                        marker = Marker(size = 12, line = Line(color = 'rgba(220, 220, 220, 0.15)', width = 0.5),
                        opacity = 0.8))
    results.append(result)

data = Data(results)
layout = Layout(showlegend = True,
                scene = Scene(xaxis = XAxis(title = 'Componente Principal 1', showline = False),
                              yaxis = YAxis(title = 'Componente Principal 2', showline = False)))

fig = Figure(data = data, layout = layout)
py.iplot(fig)
```
