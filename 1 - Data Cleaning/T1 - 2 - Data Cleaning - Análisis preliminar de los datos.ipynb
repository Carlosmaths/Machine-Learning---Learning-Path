##Podemos sustituir valores NaN, por ejemplo, por promedios. Veamos como hacerlo:

```python
# Veamos cuantos valores NaN hay en la columna "age":
data5["age"].isnull().ravel().sum()
```

```python
data1["age"].mean()
```

```python
# Ahora sustituimos los valores NaN de la columna "age", por la media de los valores existentes
data5["age"].fillna(data1["age"].mean())
```

```python
# Los cambios a nivel general no se han producido; tendríamos que hacer una reasignación
# Verifiquemos que no se han produciod dichos cambios
data5["age"].isnull().ravel().sum()
```

```python
# Está el método para completar los NaN con el valor inmediato después de éste
data5['age'].fillna(method = "ffill")
```

```python
# Contemos nuevamente
data5["age"].isnull().ravel().sum()
```

```python
# Está también el método para completar los NaN con el valor inmediato previo de éste
data5['age'].fillna(method = "backfill")
```

```python
data1.head()
```

```python
# Hagamos una función que simplifique el proceso:

def createDummies(data, category):
    dummy = pd.get_dummies(data[category], prefix = category)
    data = data.drop(category, axis = 1)
    data = pd.concat([data, dummy], axis = 1)
    return data
```

```python
# Probamos la función creada:
createDummies(data5, "sex")
```
