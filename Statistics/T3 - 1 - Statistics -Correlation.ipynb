## Trabajaremos con el coeficiente de correlación de Pearson.

* numerator = Cov(X, Y)
* var1 = Var(X)
* var2 = Var(Y)
* denominator = Raíz(var1 * var2)
* coeficiente de relación entre X e Y:
  
  r = numerator / denominator

```python
def coef_correlation(df, col1, col2):
    numerator = np.sum(((df[str(col1)] - np.mean(df[str(col1)])) * (df[str(col2)] - np.mean(df[str(col2)]))))
    variance1 = np.sum((df[str(col1)] - np.mean(df[str(col1)])) ** 2)
    variance2 = np.sum((df[str(col2)] - np.mean(df[str(col2)])) ** 2)
    denominator = np.sqrt(variance1 * variance2)
    r = numerator / denominator

    return "El coeficiente de relación entre las columnas {} y {} es: {}".format(col1, col2, r)
```

```python
coef_correlation(data_ads, "TV", "Sales")
```

```python
columns_data_ads = data_ads.columns.values
columns_data_ads
```

```python
for i in columns_data_ads:
    for j in columns_data_ads:
        print(coef_correlation(data_ads, i, j))
        print("")
```
