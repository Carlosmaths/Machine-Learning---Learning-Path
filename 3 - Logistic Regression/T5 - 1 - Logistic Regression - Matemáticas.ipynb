## La probabilidad condicional

* Cuál es la probabilidad de que un cliente compre un producto, sabiendo que es hombre?



```python
"""
Total de hombres en la observación = 125 + 121 = 246
Total de hombres que compraron = 125
P = 121 / 246
"""

121 / 246
```

* Cuál es la probabilidad de que una compra realizada, haya sido por una mujer?

```python
"""
Total de mujeres que compraron = 159
Total de personas que compraron = 159 + 121 = 280
P = 159 / 280
"""
159 / 280
```

### Probabilidades que pueden ser de interés (ver el comando): `contingency_table.astype('float').div(contingency_table.sum(axis = 1), axis = 0)`:

* Probabilidad de que siendo mujer, compre = 60%
* Probabilidad de que siendo mujer, no compre = 40%
* Probabilidad de que siendo hombre, compre = 49%
* Probabilidad de que siendo hombre, no compre = 51%

## Ratio de probabilidades:
### Es el cociente de los casos de éxito entre los casos de fracaso en el suceso estudiado y para cada grupo.

* Pm = probabilidad de comprar, siendo hombre.
* Pf = probabilidad de comprar, siendo mujer.

### Según las tablas anteriores:

* Pm = 121 / 246
* Pf = 159 / 265


## Entonces:

* Cm = Pm / (1 - Pm)
* Cf = Pf / (1 - Pf)

```python
'''
Pm = probabilidad de comprar, siendo hombre.
Pf = probabilidad de comprar, siendo mujer.

Pm = 121 / 246
Pf = 159 / 265

odds_m ~ Cm
odds_f ~ Cf
'''

pm = 121 / 246
pf = 159 / 265

odds_m = pm / (1 - pm)
odds_f = pf / (1 - pf)

print(odds_m)
print(odds_f)
```
