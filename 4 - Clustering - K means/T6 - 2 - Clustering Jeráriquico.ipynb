# Clustering Jeráriquico y Dendrogramas

* Para el dataset nos referimos con X (array de n x m puntos a clusterizar).
* Número de datos: n.
* Número de rasgos: m.
* Z: Array de enlace del cluster con información de las uniones.
* Número de clusters: k.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
np.random.seed(4711)
a = np.random.multivariate_normal([10, 0], [[3, 1], [1, 4]], size = [100,])
b = np.random.multivariate_normal([0, 20], [[3, 1], [1, 4]], size = [50,])
X = np.concatenate((a, b))
print('shape: ', X.shape)
plt.scatter(X[:, 0], X[:, 1])
plt.show()
```

##En detalle, la matriz Z:

```python
Z = linkage(X, 'ward')
```

```python
from scipy.cluster.hierarchy import cophenet
from scipy.spatial.distance import pdist
```

```python
c, coph_dist = cophenet(Z, pdist(X))
```

## Representación gráfica del dendrograma

```python
plt.figure(figsize = (25, 10))
plt.title('dendrograma del cluster jerárquico')
plt.xlabel('índice de la muestra')
plt.ylabel('distancias')
dendrogram(Z, leaf_rotation = 90, leaf_font_size = 10.0)
plt.show()
```

### Truncar el dendrograma

```python
plt.figure(figsize = (25, 10))
plt.title('dendrograma del cluster jerárquico')
plt.xlabel('índice de la muestra')
plt.ylabel('distancias')
dendrogram(Z, leaf_rotation = 90, leaf_font_size = 10.0, truncate_mode = 'lastp', p = 7,
           show_contracted = True, show_leaf_counts = True) # Atentos con el parámetro 'p'.
plt.show()
```
