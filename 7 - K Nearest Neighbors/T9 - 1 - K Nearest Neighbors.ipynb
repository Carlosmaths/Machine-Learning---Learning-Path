# Clasificador de los K vecinos más cercanos

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2)
```

```python
# clasificador = clf

clf = neighbors.KNeighborsClassifier()
```

```python
clf.fit(X_train, Y_train)
```

```python
accuracy = clf.score(X_test, Y_test)
accuracy
```
