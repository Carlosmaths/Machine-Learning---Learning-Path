```python
data_1["Sales_pred"] = 7.032594 + 0.047537 * data_1["TV"]
```

```python
# RSE = error estándar residual

data_1["RSE"] = (data_1["Sales"] - data_1["Sales_pred"]) ** 2
```

```python
# SSD = suma de los cuadrados de las diferencias

SSD = np.sum(data_1["RSE"])
```

```python
# ahora, el cálculo de RSE mediante la fórmula

RSE = np.sqrt(SSD / (len(data_1) - 2))
```

```python
RSE
```

# Regresión Lineal Múltiple:

### Posibles modelos:

* Sales ~ TV
* Sales ~ Newspaper
* Sales ~ Radio
* Sales ~ TV + Newspaper
* Sales ~ TV + Radio
* Sales ~ Newspaper + Radio
* Sales ~ TV + Newspaper + Radio

```python
# Añadir "Newspaper# al modelo existente

linear_model_2 = smf.ols(formula = "Sales ~ TV + Newspaper", data = data_1).fit()
```

```python
linear_model_2.params
```

```python
linear_model_2.pvalues
```

```python
linear_model_2.rsquared
```

```python
linear_model_2.rsquared_adj
```

```python
# Atentos al doble corchete; se está generando un subdataset, del data set original.

sales_pred_2 = linear_model_2.predict(data_1[["TV", "Newspaper"]])
```

```python
sales_pred_2
```

```python
SSD = np.sum((data_1["Sales"] - sales_pred_2) ** 2)
```

```python
SSD
```

```python
RSE = np.sqrt(SSD / (len(data_1) - 2 - 1))
```

```python
RSE
```

```python
error = RSE / data_1['Sales'].mean()
error
```

```python
linear_model_2.summary()
```

```python
# Añadir "Radio" al modelo existente con "TV"

linear_model_3 = smf.ols(formula = "Sales ~ TV + Radio", data = data_1).fit()
```

# Multicolinealidad

### Factor Inflación de la Varianza

* VIF = 1: Las variables no están correlacionadas.
* VIF < 5: Las variables tienen una correlación moderada y pueden seguir en el modelo.
* VIF >= 5: Las variables están altamente correlacionadas y deben desaparecer del modelo.

```python
# Newspaper ~ TV + Radio --> R^2 VIF = 1 / (1 - R^2)

lm_newspaper = smf.ols(formula = "Newspaper ~ TV + Radio", data = data_1).fit()
rsquared_newspaper = lm_newspaper.rsquared
VIF_1 = 1 / (1 - rsquared_newspaper)
VIF_1
```

```python
# TV ~ Newspaper + Radio --> R^2 VIF = 1 / (1 - R^2)

lm_TV = smf.ols(formula = "TV ~ Newspaper + Radio", data = data_1).fit()
rsquared_TV = lm_TV.rsquared
VIF_2 = 1 / (1 - rsquared_TV)
VIF_2
```

```python
# Radio ~ TV + Newspaper --> R^2 VIF = 1 / (1 - R^2)

lm_radio = smf.ols(formula = "Radio ~ TV + Newspaper", data = data_1).fit()
rsquared_radio = lm_radio.rsquared
VIF_3 = 1 / (1 - rsquared_radio)
VIF_3
```

# Conclusión, no existe alto grado de colinealidad entre las tres variables. Por lo tanto, el modelo de Sales a partir de las tres variables en cuestión, resulta apto.
