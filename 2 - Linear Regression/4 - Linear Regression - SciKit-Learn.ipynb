#Regresión Lineal con Python
##El paquete `scikit-learn` para la regresión lineal y selección de rasgos.

```python
# RFE = Recursive Feature Elimination (Eliminación de funciones recursivas)
# SVR = Support Vector Regression (Soporte Vectorial de Regresión)

from sklearn.feature_selection import RFE
from sklearn.svm import SVR
import numpy as np
import pandas as pd
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
data = pd.read_csv("/content/drive/MyDrive/Machine Learning - Python/Datasets - Machine Learning/ads/Advertising.csv")
```

```python
data.head()
```

```python
feature_cols = ["TV", "Radio", "Newspaper"]
```

```python
X = data[feature_cols]
Y = data["Sales"]
```

```python
estimator = SVR(kernel = "linear")
selector_0 = RFE(estimator, n_features_to_select = 2, step = 1)
selector = selector_0.fit(X, Y)
```

```python
selector.support_
```

```python
selector.ranking_
```

```python
from sklearn.linear_model import LinearRegression
```

```python
X_pred = X[["TV", "Radio"]]
```

```python
lm = LinearRegression()
```

```python
lm.fit(X_pred, Y)
```

```python
lm.intercept_
```

```python
lm.coef_
```

```python
# Para calcular el r^2 ajustado:
lm.score(X_pred, Y)
```
